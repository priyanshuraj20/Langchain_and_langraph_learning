#**Chains(pipeline):A Chain is a sequence of components connected together to perform a task.** hum multiple components ko connect karte hain taaki ek component ka output next component ka input ban sake.


Langchain uses LCEL (LangChain Expression Language).   |(pipe operator) is used to connect multiple components in a chain. The output of one component is passed as input to the next component in the chain.

```python
#without chain:

from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

llm = init_chat_model("gpt-5-mini")

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Explain {topic} in simple words.")
])

# Step 1: Prepare input variables
input_data = {
    "role": "Python teacher",
    "topic": "Decorators"
}

# Step 2: Generate formatted messages
formatted_prompt = prompt_template.invoke(input_data)

# Step 3: Send messages to LLM
response = llm.invoke(formatted_prompt)

# Step 4: Extract response
print(response.content)



#chain:
#ek chiz yaad rakho ek component ka output uske aage wala ka input mein jata 



from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = init_chat_model("gpt-5-mini")

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Explain {topic} in simple words.")
])

# Parser ka kaam hai LLM ke raw output ko required format ya structure mein convert karna.
parser = StrOutputParser()

chain = prompt | llm | parser

response = chain.invoke({
    "role": "Python teacher",
    "topic": "Decorators"
})

print(response.content)
```
