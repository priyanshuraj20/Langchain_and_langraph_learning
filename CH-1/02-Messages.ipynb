```python
# Messages
from langchain_core.messages import (
    HumanMessage,
    SystemMessage,
    AIMessage
)

from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm_openai = init_chat_model("gpt-5-mini")

my_messages = [
    SystemMessage(
        content="You are a helpful assistant who always answers in a fun way"
    ),
    HumanMessage(
        content="Give me an interesting fun fact"
    )
]

response = llm_openai.invoke(my_messages)

print(response.content)
```

```python
#overAll    connect ChatPromptTemplate with LLM :


from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

llm = init_chat_model("gpt-5-mini")

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Explain {topic} with an example.")
])

chain = prompt_template | llm

response = chain.invoke({
    "role": "Python teacher",
    "topic": "Decorators"
})

print(response.content)
```
