# **Guiding in Prompts**


```python
import os
import load_dotenv
load_dotenv()


if os.environ.get("OPENAI_API_KEY"):
    print("Api key variable exist")
else:
    raise ValueError("OPENAI_API_KEY not found")
```

```python
# Now we will guide the llm model to genrate the output in specific schema


from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

llm = init_chat_model("gpt-5-mini")

dynamic_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", """
        Explain {topic} with an example.
        Return the output in key-value pairs.
    """)
])

chain = dynamic_template | llm

response = chain.invoke({
    "role": "Python teacher",
    "topic": "Decorators"
})

print(response.content)
```

# **Using Pydantic Models for Structured Output**
Langchain provides with_structured_output() to guide the LLM to generate structured output. This is useful when you want the LLM to return data in a specific format, such as JSON or a Python dictionary.

```python
from langchain.chat_models import init_chat_model
from pydantic import BaseModel, Field
#baseModel: is a class that we use the create our schema 

llm = init_chat_model("gpt-5-mini")

class llm_schema(BaseModel):
    definition: str = Field(description="What is the topic?")
    example: str = Field(description="A practical example")
    use_case: str = Field(description="Where is it used?")

structured_llm = llm.with_structured_output(llm_schema)
# structured_llm

response = structured_llm.invoke(
    "Explain Python pydantic"
) 

print(response)
```
