```python
# Now we will guide the llm model to genrate the output in specific schema


from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

llm = init_chat_model("gpt-5-mini")

dynamic_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", """
        Explain {topic} with an example.
        Return the output in key-value pairs.
    """)
])

chain = dynamic_template | llm

response = chain.invoke({
    "role": "Python teacher",
    "topic": "Decorators"
})

print(response.content)
```

# **Using Pydantic Models for Structured Output**
Langchain provides with_structured_output() to guide the LLM to generate structured output. This is useful when you want the LLM to return data in a specific format, such as JSON or a Python dictionary.

```python
from langchain.chat_models import init_chat_model
from pydantic import BaseModel, Field
#baseModel: is a class that we use the create our schema 

llm = init_chat_model("gpt-5-mini")

class llm_schema(BaseModel):
    definition: str = Field(description="What is the topic?")
    example: str = Field(description="A practical example")
    use_case: str = Field(description="Where is it used?")

structured_llm = llm.with_structured_output(llm_schema)
# structured_llm

response = structured_llm.invoke(
    "Explain Python pydantic"
) 

print(response)
```

# TypedDict : TypedDict is a Python feature used to define the expected structure and types of keys in a dictionary. It provides static type checking but does not perform runtime data validation.

TypedDict Python ka ek feature hai jo dictionary ka structure define karta hai.

Matlab dictionary mein:
- Kaunsi keys honi chahiye.
- Har key ki expected value ka type kya hona chahiye.

```python
from typing import TypedDict

class Student(TypedDict):
    name: str
    age: int
    course: str

student: Student = {
    "name": "Rahul",
    "age": 21,
    "course": "B.Tech"
}

# Yahan Student ek normal Python class jaisa object nahi hai. Ye dictionary ka expected structure define karta hai.


#Important: TypedDict runtime par automatically data validate nahi karta. Ye primarily static type checking aur editor support ke liye hai.


#LangChain mein jab hum RunnableLambda, RunnableParallel jaise components use karte hain, toh hum input/output ka expected structure define kar sakte hain.


# A static type checker is a tool that analyzes your code for type mismatches and data structure bugs before you actually run the program. In Python, this means verifying your type hints (like int, str, or TypedDict) automatically as you write code or during automated builds.

from typing import TypedDict
from langchain_core.runnables import RunnableLambda

class InputData(TypedDict):
    name: str
    age: int

def process_student(data: InputData):
    return f"{data['name']} is {data['age']} years old."

chain = RunnableLambda(process_student)

result = chain.invoke({
    "name": "Rahul",
    "age": 21
})

print(result)
```
