# Custom Runnable

Iska main purpose hai apne khud ke Python functions ko LangChain chain ka part banana.

Runnable? Iska main purpose hai apne khud ke Python functions ko LangChain chain ka part banana.

invoke() – input process karke output deta hai.
batch() – multiple inputs process karta hai.
stream() – output ko chunks mein deta hai.

```python
RunnableLambda : Maan le humein LLM ke response ko uppercase mein convert karna hai.yeh kisi output ko as input kisi template mein dene se phale dictionary mein convert karna ho toh RunnableLamba hi use karte

```

```python

from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableLambda
from langchain_core.output_parsers import StrOutputParser

llm = init_chat_model("gpt-5-mini",temperature = 0)


# Task 5: Create two prompt templates

# First prompt: Generate an explanation
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Explain {topic} in simple words.")
])

# Second prompt: Generate a LinkedIn post
prompt_post = ChatPromptTemplate.from_messages([
    ("system", "You are a social media post generator."),
    ("human", "Create a post for LinkedIn using this information: {text}")
])


# Custom function 1: Convert output into uppercase
def make_uppercase(text: str) -> str:
    return text.upper()


# Custom function 2: Convert string into dictionary
def make_dictionary(text: str) -> dict:
    return {"text": text}


# These are normal Python functions.
# To execute them sequentially inside a LangChain chain,
# we convert them into RunnableLambda.

uppercase_runnable = RunnableLambda(make_uppercase)
dictionary_runnable = RunnableLambda(make_dictionary)


# Create a sequential chain
chain = (
    prompt
    | llm
    | StrOutputParser()
    | uppercase_runnable
    | dictionary_runnable
    | prompt_post
    | llm
    | StrOutputParser()
)


# Execute the complete chain
response = chain.invoke({
    "role": "Python teacher",
    "topic": "Decorators"
})

print(response)
```
