# LangGraph Basics :
Langgraph is framework for building stateful, multi-step Ai workflows(espcially agentic workflows) using a graph-based approach. It allows developers to create complex AI applications by defining nodes and edges that represent different steps in the workflow.

It allows you to define nodes that represent different tasks or actions, and edges that represent the flow of data or control between those tasks. This makes it easy to visualize and manage the workflow, as well as to add or remove steps as needed.

Key features of LangGraph include:
- **Graph-based Workflow**: Define workflows as directed graphs, making it easy to visualize and manage complex processes.
- **Stateful Execution**: Maintain state across different steps in the workflow, allowing for more complex and dynamic behavior.
- **Multi-step Workflows**: Support for workflows that involve multiple steps, enabling the creation of sophisticated AI applications.  

Directed cycles are allowed in LangGraph, which means that you can create workflows that loop back to previous steps if needed. This is particularly useful for tasks that require iterative processing or feedback loops.


This kind of workflow is exactly what LangGraph helps you build.


LangGraph is not an LLM, it is not a model, and it does not independently make an AI intelligent.

It provides the architecture and control flow to build complex AI applications.

# ReAct Agent in Langraph :
Dekho jab mein langchain padha tha, mujhe ReAct agent ka concept samajh aaya. ReAct agent ek aisa agent hai jo reasoning aur acting dono karta hai. Ye agent apne observations ke basis par reasoning karta hai aur phir us reasoning ke basis par action leta hai. But langchain mein HITL(human in the loop) ka concept inbuilt nahi provide karta but in LangGraph, **HITL** ka concept inbuilt provide karta hai. Iska matlab ye hai ki aap apne agent ke decision making process mein human feedback ko include kar sakte ho. Ye feature bahut useful hai jab aapko complex tasks perform karne hote hain jahan par human judgment ki zarurat hoti hai.




**"Human-in-the-Loop is a mechanism that allows human intervention in an AI workflow, particularly before sensitive or critical actions. In LangGraph, interrupts can pause graph execution, wait for human input, and resume execution after approval or modification."**


LangGraph ka interrupt() execution pause karta hai, checkpointer state persist karta hai, aur Command(resume=...) execution continue karne deta hai.
